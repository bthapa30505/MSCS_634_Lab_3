```python
# Step 1: Load and Prepare the Dataset 
# Create a new Jupyter Notebook and include a Markdown cell at the top with your name, course title, and lab assignment title. 
# Load the Wine Dataset from sklearn. 
# Perform basic data exploration to examine the feature details and class distribution. 
# Standardize the dataset using z-score normalization to ensure consistent scaling across features. 
```

```python
#basic imports at the top
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn_extra.cluster import KMedoids
from sklearn.metrics import silhouette_score, adjusted_rand_score
import seaborn as sns
```

```python
#loading the wine dataset
wine = load_wine()
X = wine.data
y = wine.target

print("Dataset shape:", X.shape)
print("\nFeature names:")
print(wine.feature_names)
print("\nClass distribution:")
print(f"Class 0: {np.sum(y == 0)} samples")
print(f"Class 1: {np.sum(y == 1)} samples")
print(f"Class 2: {np.sum(y == 2)} samples")
print("\nTotal samples:", len(y))
```

```python
# performing the basic data exploration
df = pd.DataFrame(X, columns=wine.feature_names)
df['target'] = y

print("Dataset Info:")
print(df.info())
print("\nFirst few rows:")
print(df.head())
print("\nStatistical summary:")
print(df.describe())
```

```python
# Standardize the dataset using z-score normalization
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Original data shape:", X.shape)
print("Scaled data shape:", X_scaled.shape)
print("\nScaled data statistics:")
print(f"Mean: {np.mean(X_scaled, axis=0)[:5]}...")  # for first 5 features
print(f"Std: {np.std(X_scaled, axis=0)[:5]}...")   # for first 5 features
```

```python
# Step 2: Implement K-Means Clustering 
# Implement the K-Means algorithm with k = 3 (since the Wine dataset has 3 classes). 
# Train the model and obtain the cluster labels. 
# Calculate the following performance metrics: 
# Silhouette Score (to measure how well-defined the clusters are). 
# Adjusted Rand Index (ARI) (to compare the clusters with the actual class labels).
```

```python
# Implement K-Means clustering with k=3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X_scaled)
kmeans_centroids = kmeans.cluster_centers_
print(f"Number of clusters: {len(np.unique(kmeans_labels))}")
print(f"\nCluster assignments:")
for i in range(3):
    print(f"Cluster {i}: {np.sum(kmeans_labels == i)} samples")
```

```python
# Calculate performance metrics for K-Means
kmeans_silhouette = silhouette_score(X_scaled, kmeans_labels)
kmeans_ari = adjusted_rand_score(y, kmeans_labels)

print("K-Means Performance Metrics:")
print(f"Silhouette Score: {kmeans_silhouette:.4f}")
print(f"Adjusted Rand Index (ARI): {kmeans_ari:.4f}")
```

```python
# Step 3: Implement K-Medoids Clustering 
# Implement the K-Medoids algorithm with k = 3. 
# Train the model and obtain the cluster labels. 
# Calculate the same performance metrics as in Step 2. 
# Silhouette Score 
# Adjusted Rand Index (ARI). 
```

```python
# Implement K-Medoids clustering with k=3
kmedoids = KMedoids(n_clusters=3, random_state=42, method='pam')
kmedoids_labels = kmedoids.fit_predict(X_scaled)
kmedoids_medoids = kmedoids.cluster_centers_

print("K-Medoids clustering completed!")
print(f"Number of clusters: {len(np.unique(kmedoids_labels))}")
print(f"\nCluster assignments:")
for i in range(3):
    print(f"Cluster {i}: {np.sum(kmedoids_labels == i)} samples")
```

```python
# Calculate performance metrics for K-Medoids
kmedoids_silhouette = silhouette_score(X_scaled, kmedoids_labels)
kmedoids_ari = adjusted_rand_score(y, kmedoids_labels)
print("K-Medoids Performance Metrics:")
print(f"Silhouette Score: {kmedoids_silhouette:.4f}")
print(f"Adjusted Rand Index (ARI): {kmedoids_ari:.4f}")
```

```python
# Step 4: Visualize and Compare Results 
# Create side-by-side scatter plots for both the K-Means and K-Medoids clusters. 
# Mark the cluster centroids/medoids on the plots for visual clarity. 
# Provide a brief analysis comparing the results: 
# Which algorithm produced better-defined clusters? 
# What differences do you observe in cluster shapes or positioning? 
# Based on your results, explain when K-Means or K-Medoids may be preferable. 
```

```python
# Reduce dimensions for visualization using PCA
from sklearn.decomposition import PCA

# Apply PCA to reduce to 2D for visualization
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

# Debug: Check centroids and medoids shapes
print("K-Means centroids shape:", kmeans_centroids.shape)
print("K-Medoids medoids shape:", kmedoids_medoids.shape)
print("Number of unique K-Means clusters:", len(np.unique(kmeans_labels)))
print("Number of unique K-Medoids clusters:", len(np.unique(kmedoids_labels)))

# Transform centroids/medoids to PCA space
kmeans_centroids_pca = pca.transform(kmeans_centroids)
kmedoids_medoids_pca = pca.transform(kmedoids_medoids)

print(f"\nK-Means centroids in PCA space shape: {kmeans_centroids_pca.shape}")
print(f"K-Medoids medoids in PCA space shape: {kmedoids_medoids_pca.shape}")
print(f"\nPCA explained variance ratio: {pca.explained_variance_ratio_}")
print(f"Total variance explained: {sum(pca.explained_variance_ratio_):.4f}")
```

```python
# Creating side-by-side scatter plots
# Increased width to accommodate legend outside plot
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Ensure we only have 3 centroids/medoids
print(f"K-Means centroids_pca shape: {kmeans_centroids_pca.shape}")
print(f"K-Medoids medoids_pca shape: {kmedoids_medoids_pca.shape}")
print(f"\nK-Means centroids coordinates (PCA space):")
for i, centroid in enumerate(kmeans_centroids_pca):
    print(f"  Centroid {i}: ({centroid[0]:.4f}, {centroid[1]:.4f})")
print(f"\nK-Medoids medoids coordinates (PCA space):")
for i, medoid in enumerate(kmedoids_medoids_pca):
    print(f"  Medoid {i}: ({medoid[0]:.4f}, {medoid[1]:.4f})")

# Ensure exactly 3 clusters
assert kmeans_centroids_pca.shape[0] == 3, f"Expected 3 centroids, got {kmeans_centroids_pca.shape[0]}"
assert kmedoids_medoids_pca.shape[0] == 3, f"Expected 3 medoids, got {kmedoids_medoids_pca.shape[0]}"

# Explicitly use only the first 3 centroids/medoids (in case there are more)
kmeans_centroids_plot = kmeans_centroids_pca[:3]
kmedoids_medoids_plot = kmedoids_medoids_pca[:3]

# K-Means plot
ax1 = axes[0]
scatter1 = ax1.scatter(X_pca[:, 0], X_pca[:, 1], c=kmeans_labels, cmap='viridis', 
                      alpha=0.6, s=50, edgecolors='black', linewidth=0.5)
# Plot exactly 3 centroids (note: legend will show an additional marker symbol)
ax1.scatter(kmeans_centroids_plot[:, 0], kmeans_centroids_plot[:, 1], 
           c='red', marker='X', s=300, label='Centroids (3)', 
           edgecolors='black', linewidth=2, zorder=10)
ax1.set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)', fontsize=12)
ax1.set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)', fontsize=12)
ax1.set_title(f'K-Means Clustering\nSilhouette: {kmeans_silhouette:.4f}, ARI: {kmeans_ari:.4f}', 
             fontsize=14, fontweight='bold')
# Position legend outside the plot to avoid confusion
ax1.legend(loc='upper left', bbox_to_anchor=(1.02, 1))
ax1.grid(True, alpha=0.3)

# K-Medoids plot
ax2 = axes[1]
scatter2 = ax2.scatter(X_pca[:, 0], X_pca[:, 1], c=kmedoids_labels, cmap='viridis', 
                      alpha=0.6, s=50, edgecolors='black', linewidth=0.5)
# Plot exactly 3 medoids (note: legend will show an additional marker symbol)
ax2.scatter(kmedoids_medoids_plot[:, 0], kmedoids_medoids_plot[:, 1], 
           c='red', marker='X', s=300, label='Medoids (3)', 
           edgecolors='black', linewidth=2, zorder=10)
ax2.set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)', fontsize=12)
ax2.set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)', fontsize=12)
ax2.set_title(f'K-Medoids Clustering\nSilhouette: {kmedoids_silhouette:.4f}, ARI: {kmedoids_ari:.4f}', 
             fontsize=14, fontweight='bold')
# Position legend outside the plot to avoid confusion
ax2.legend(loc='upper left', bbox_to_anchor=(1.02, 1))
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Create comparison plot with actual labels for reference
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Actual labels
ax0 = axes[0]
scatter0 = ax0.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='viridis', 
                      alpha=0.6, s=50, edgecolors='black', linewidth=0.5)
ax0.set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)', fontsize=12)
ax0.set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)', fontsize=12)
ax0.set_title('Actual Wine Classes (Ground Truth)', fontsize=14, fontweight='bold')
ax0.grid(True, alpha=0.3)

# K-Means plot
ax1 = axes[1]
scatter1 = ax1.scatter(X_pca[:, 0], X_pca[:, 1], c=kmeans_labels, cmap='viridis', 
                      alpha=0.6, s=50, edgecolors='black', linewidth=0.5)
ax1.scatter(kmeans_centroids_pca[:, 0], kmeans_centroids_pca[:, 1], 
           c='red', marker='X', s=300, label='Centroids', edgecolors='black', linewidth=2)
ax1.set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)', fontsize=12)
ax1.set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)', fontsize=12)
ax1.set_title(f'K-Means\nSilhouette: {kmeans_silhouette:.4f}, ARI: {kmeans_ari:.4f}', 
             fontsize=14, fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# K-Medoids plot
ax2 = axes[2]
scatter2 = ax2.scatter(X_pca[:, 0], X_pca[:, 1], c=kmedoids_labels, cmap='viridis', 
                      alpha=0.6, s=50, edgecolors='black', linewidth=0.5)
ax2.scatter(kmedoids_medoids_pca[:, 0], kmedoids_medoids_pca[:, 1], 
           c='red', marker='X', s=300, label='Medoids', edgecolors='black', linewidth=2)
ax2.set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)', fontsize=12)
ax2.set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)', fontsize=12)
ax2.set_title(f'K-Medoids\nSilhouette: {kmedoids_silhouette:.4f}, ARI: {kmedoids_ari:.4f}', 
             fontsize=14, fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Create a comparison table
comparison_df = pd.DataFrame({
    'Algorithm': ['K-Means', 'K-Medoids'],
    'Silhouette Score': [kmeans_silhouette, kmedoids_silhouette],
    'Adjusted Rand Index': [kmeans_ari, kmedoids_ari]
})

print("Performance Comparison:")
print(comparison_df.to_string(index=False))
print("\nMetrics:")
print("-"*50)
print(f"\nSilhouette Score:")
print(f"  K-Means:   {kmeans_silhouette:.4f}")
print(f"  K-Medoids: {kmedoids_silhouette:.4f}")
if kmeans_silhouette > kmedoids_silhouette:
    print(f"  → K-Means has better-defined clusters")
else:
    print(f"  → K-Medoids has better-defined clusters")

print(f"\nAdjusted Rand Index (ARI):")
print(f"  K-Means:   {kmeans_ari:.4f}")
print(f"  K-Medoids: {kmedoids_ari:.4f}")
if kmeans_ari > kmedoids_ari:
    print(f"  → K-Means better matches the ground truth labels")
else:
    print(f"  → K-Medoids better matches the ground truth labels")
```

## Analysis and Comparison

The analysis comparing K-Means and K-Medoids clustering reveals distinct differences in performance, robustness, and applicability. Based on the silhouette and Adjusted Rand Index (ARI) scores, K-Means typically achieves slightly higher silhouette values, indicating better-defined clusters when data is well-distributed and free of significant outliers. However, K-Medoids often shows higher ARI scores, suggesting stronger agreement with the true wine classes, especially when the dataset contains noise or irregular cluster shapes. Visualizations further highlight that K-Means forms more compact and spherical clusters, while K-Medoids captures irregular boundaries more effectively by using actual data points as centers. Given these characteristics, K-Means is generally preferred for large, continuous datasets where computational efficiency matters, whereas K-Medoids is more suitable for smaller, noisy datasets or when interpretability and robustness to outliers are priorities.
